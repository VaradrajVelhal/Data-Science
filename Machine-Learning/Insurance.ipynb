# Data cleaning and preprocessing

```python
df_cleaned = df.copy()
df_cleaned.head()
df_cleaned.shape
```

```python
df_cleaned.drop_duplicates(inplace=True)
```

```python
df_cleaned.shape
```

```python
df_cleaned.dtypes
```

```python
df_cleaned['sex'].value_counts()
```

```python
df_cleaned['sex'] = df_cleaned['sex'].map({'male' : 0, 'female' : 1})
```

```python
df_cleaned.head()
```

```python
df_cleaned['smoker'].value_counts()
```

```python
df_cleaned['smoker'] = df_cleaned['smoker'].map({'no': 0, 'yes' : 1})
```

```python
df_cleaned.head()
```

```python
df_cleaned.rename(columns={'sex' : 'is_female', 'smoker' : 'is_smoker'}, inplace = True)
df_cleaned.head()
```

```python
df_cleaned['region'].value_counts()
```

```python
df_cleaned = pd.get_dummies(df_cleaned, columns=['region'], drop_first=True)
df_cleaned.head()
```

```python
df_cleaned = df_cleaned.astype(int)
df_cleaned.head()
```

# Feature engineering and extraction

```python
df_cleaned['bmi_category'] = pd.cut(df_cleaned['bmi'], bins=[0, 18.5, 24.9, 29.9, float(' inf')], labels=['Underweight', 'Normal weight', 'Overweight', 'Obese'])
```

```python
df_cleaned.head()
```

```python
df_cleaned = pd.get_dummies(df_cleaned, columns=['bmi_category'], drop_first=True)
```

```python
df_cleaned = df_cleaned.astype(int)
```

```python
df_cleaned.head()
```

```python
from sklearn.preprocessing import StandardScaler
cols = ['age', 'bmi', 'children']
scaler = StandardScaler()
df_cleaned[cols] = scaler.fit_transform(df_cleaned[cols])
df_cleaned.head()
df_cleaned.columns = df_cleaned.columns.str.replace(' ', '_')
df_cleaned.columns
```

```python
from scipy.stats import pearsonr

# ----------------------------------
# Pearson Correlation Calculation
# ----------------------------------

# List of features to check against target
selected_features = [
    'age', 'bmi', 'children', 'is_female', 'is_smoker',
    'region_northwest', 'region_southeast', 'region_southwest',
    'bmi_category_Normal_weight', 'bmi_category_Overweight', 'bmi_category_Obese'
]

correlations = {
    feature: pearsonr(df_cleaned[feature], df_cleaned['charges'])[0]
    for feature in selected_features
}
correlation_df = pd.DataFrame(list(correlations.items()), columns=['Feature', 'Pearson Correlation'])
correlation_df.sort_values(by='Pearson Correlation', ascending=False)
```

```python
cat_features = [
    'is_female', 'is_smoker',
    'region_northwest', 'region_southeast', 'region_southwest',
    'bmi_category_Normal_weight', 'bmi_category_Overweight', 'bmi_category_Obese'
]
```

```python
from scipy.stats import chi2_contingency
import pandas as pd

alpha = 0.05

df_cleaned['charges_bin'] = pd.qcut(df_cleaned['charges'], q=4, labels=False)
chi2_results = {}

for col in cat_features:
    contingency = pd.crosstab(df_cleaned[col], df_cleaned['charges_bin'])
    chi2_stat, p_val, _, _ = chi2_contingency(contingency)
    decision = 'Reject Null (Keep Feature)' if p_val < alpha else 'Accept Null (Drop Feature)'
    chi2_results[col] = {
        'chi2_statistic': chi2_stat,
        'p_value': p_val,
        'Decision': decision
    }

chi2_df = pd.DataFrame(chi2_results).T
chi2_df = chi2_df.sort_values(by='p_value')
chi2_df
```
