### Loading the Libraries and Dataset

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import warnings
warnings.filterwarnings('ignore')

df = pd.read_csv('IPL.csv')
df.head()
```

**Now, Here comes some Basic Questions**

1. Which team won the most matches?

```python
match_wins = df['match_winner'].value_counts()
sns.barplot(y=match_wins.index, x=match_wins.values, palette='viridis')
plt.title('Number of Matches Won by Each Team')
```

2. Toss Decision Trends

```python
sns.countplot(x = df['toss_decision'], palette='Set2')
```

3. Toss Winner vs Match Winner

```python
count = df[df['toss_winner'] == df['match_winner']]['match_id'].count()
percentage = (count * 100) / df.shape[0]
print(f"Teams that won the toss also won the match {percentage:.2f}% of the time.")
```

```python
sns.countplot(x = df['won_by'], palette='Set3')
plt.title("Winning Margin Trends")
```

###  Key Player Performances

1 Most "Player of the Match" Awards

```python
count = df['player_of_the_match'].value_counts().head(10)
count
```

```python
sns.barplot(y = count.index, x = count.values, palette='mako')
plt.title("Top 10 Players with Most Player of the Match Awards")
```

2 Top Scorers

```python
df.head()
```

```python
high = df.groupby('top_scorer')['highscore'].sum().sort_values(ascending=False).head(2)
high
```

```python
high.plot(kind='bar', color=['blue', 'orange'])
```

10 Best Bowling Figures

```python
df['highest_wickets'] = df['best_bowling_figure'].apply(lambda x: x.split('--')[0])
df['highest_wickets'] = df['highest_wickets'].astype(int)
df
```

```python
top_bowlers = df.groupby('best_bowling')['highest_wickets'].sum().sort_values(ascending=False).head(10)
top_bowlers.plot(kind='barh')
```
